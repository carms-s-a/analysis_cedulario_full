# Named Entity Recognition (NER)

```
# Definir función para normalizar
```

```
def normalize(text):
	text=str(text)
	text=text.lower()
	text=text.strip()
	text = re.sub(r"\s+", " ", text)
	text = re.sub(r"[^\w\sáéíóúñü]", "", text)
	return text
```

```
# Abrir y pre-procesar la transcripción del cedulario en .txt
```

```
with open("data/complete_transcription.txt", "r", encoding="utf-8") as f:
	texto=f.read()
doc=nlp(texto)
```

```
# Extraer entidades
```

```
entidades=set()
for ent in doc.ents:
	ent_norm=normalize(ent.text)
	entidades.add((ent.text, ent.label_))
```

```
# Guardar entidades en documento .csv para su posterior normalización
```

```
data = [(ent.text.lower(), ent.label_) for ent in doc.ents]
df = pd.DataFrame(data, columns=["entidad_normalizada", "etiqueta"])
df.to_csv("entidades.csv", index=False, encoding="utf-8")
```
